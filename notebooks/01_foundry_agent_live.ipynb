## PRESENTATION step C — One consented real-model question

**Exact import:** `presentation_live.run_question(settings, text, role,
state_dir, authorized=True)` returns a dictionary with observed run evidence.
Inputs are a fictional policy question, the synthetic `employee` role label,
a new private local state directory and the previously pinned dedicated agent
configuration. The role is an application label, not authentication. Unsupported

The version check exists before invocation to reject absent/mutable pins.
It proves only that a non-latest value was supplied; the live implementation
must verify the actual remote definition. Functions execute in the client only
after pre-execution policy, with a bounded serial loop: at most three model requests and six proposed tool
calls per scene/question, with `tool_choice=none` on the third/final round. Output below reports
actual status/export fields when returned, never invented answer scores.
UNREPORTED means that field was absent. The full result remains private for
evidence inspection. No records-request approval is supplied automatically.

```python
RUN_AZURE_LIVE = False
if RUN_AZURE_LIVE:
    from pathlib import Path
    from uuid import uuid4

    from foundry_security_accelerator.config import Settings
    from foundry_security_accelerator.presentation_live import run_question

    if input(
        "Authorize one bounded synthetic presentation question and cost? "
        "Type RUN PRESENTATION QUESTION: "
    ) != ("RUN PRESENTATION QUESTION"):
        raise PermissionError("Presentation invocation not authorized")
    question_root = Path.cwd() if Path("pyproject.toml").exists() else Path.cwd().parent
    question_settings = Settings(mode="azure", _env_file=question_root / ".env")
    AUTHORIZE_TELEMETRY_EXPORT = False  # Independent opt-in after prior review
    if question_settings.export_monitor or question_settings.otlp_endpoint:
        if not AUTHORIZE_TELEMETRY_EXPORT:
            raise PermissionError(
                "Configured egress requires separate reviewed approval and opt-in"
            )
        if input(
            "Confirm prior review of configured telemetry destination/content/cost. "
            "Type APPROVED TELEMETRY EGRESS: "
        ) != ("APPROVED TELEMETRY EGRESS"):
            raise PermissionError("Telemetry egress not authorized")
    if (
        not question_settings.presentation_agent_version
        or str(question_settings.presentation_agent_version).lower() == "latest"
    ):
        raise ValueError(
            "Pin the returned dedicated presentation version privately before invoking"
        )
    question_state = question_root / ".fsa" / "notebooks" / str(uuid4())
    presentation_answer = run_question(
        question_settings,
        text="Summarize the synthetic retention policy and cite the approved policy section.",
        role="employee",
        state_dir=question_state,
        authorized=True,
    )
    print(
        {
            "status": presentation_answer.get("status", "UNREPORTED"),
            "telemetry_export": presentation_answer.get("trace_export_status", "UNREPORTED"),
            "identity_boundary": "SYNTHETIC_ROLE_LABELS_NOT_ENTRA_OBO",
        }
    )
else:
    print("Presentation question NOT_EXECUTED")
```

## Live report contract and bounded execution

`presentation_live.run_question(...)` returns observed application evidence:
`status`, `model_calls`, `tool_calls`, `citations`, `quarantined_sections`,
`correlation_id`, `audit`, `audit_head`, `audit_verified`, `trace_path` and
`trace_export_status`. These field names are specific to PRESENTATION:
do not substitute the equipment API's `traces` or `telemetry_export`.
An expected denial can be a truthful safe outcome; an exception, failed
structural check or missing field must not be relabeled successful.

`run_scenes(...)` returns `status`, `scenes` and `manifest_path`.
Per-scene structural checks are not cloud model-judge scores. The runtime
is serial, bounded to three model requests and six proposed tool calls per
scene/question; the third/final round uses `tool_choice=none`.
Proposed calls can be denied before execution. Six scenes therefore do
not imply six calls or identical tool sequences, and repeated runs repeat
cost. Plan the initial verification within 18 model calls total.

All examples are opt-in. If already-configured application export is
intended, its destination, content, retention and cost require prior
independent review; then deliberately set the nested
`AUTHORIZE_TELEMETRY_EXPORT` flag and confirm its separate prompt.
No notebook configures export or changes native account logging.
A returned export status is still not ingestion. Query actual correlated
rows separately; inspect native Agent 365 activity independently.
